```python
df["Rating"]=df["Rating"].str.extract(r"(\d+\.+\d+)")
df["Rating"]=df["Rating"].astype(float)
```

```python
df["bought_in_last_month"] = df["bought_in_last_month"].astype(str)

# Remove common suffixes like "+ bought in past month" and any optional leading "+"
df["bought_in_last_month"] = df["bought_in_last_month"].str.replace(r"\s*\+?\s*bought in past month", "", regex=True)

def convert_bought(value):
    value = str(value).strip()

    # Handle cases like 'nan' strings or empty strings after cleaning
    if not value or value.lower() == 'nan':
        return np.nan

    # Handle 'K' for thousands
    if "K" in value:
        value = value.replace("K", "")
        # Use pd.to_numeric with errors='coerce' to turn unparseable values into NaN
        return pd.to_numeric(value, errors='coerce') * 1000
    else:
        # For other numeric strings, convert directly
        return pd.to_numeric(value, errors='coerce')

df["bought_in_last_month"] = df["bought_in_last_month"].apply(convert_bought)
df["bought_in_last_month"].head(220)
```

```python
df['bought_in_last_month']
```

```python
df["Listed_price"]=df["Listed_price"].replace("No discount",np.nan)
```

```python
df["Listed_price"] = (
    df["Listed_price"]
    .str.replace("$", "", regex=False)
)
df["Listed_price"] = (
    df["Listed_price"]
    .str.replace(",", "", regex=False)
)
```

```python
df["Listed_price"] = pd.to_numeric(
    df["Listed_price"],
    errors="coerce"
)
```

```python
df["is_best_seller"] = (
    df["is_best_seller"] == "Best Seller"
)
df["is_best_seller"].dtype
```

```python
df["is_sponsored"] = df["is_sponsored"] == "Sponsored"
df["is_sponsored"].dtype
```

```python
df["is_couponed"] = df["is_couponed"] != "No Coupon"
```

```python
df["collected_at"] = pd.to_datetime(
    df["collected_at"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)
```
